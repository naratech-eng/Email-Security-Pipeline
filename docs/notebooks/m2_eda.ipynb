```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
email_df = pd.read_csv("../data/raw/phishing-email/phishing_email.csv")

url_df = pd.read_csv("../data/raw/url/malicious_phish.csv")
```

```python
print("EMAIL DATASET")
print(email_df.head())

print("\nURL DATASET")
print(url_df.head())
```

```python
print("Email Dataset Shape:")
print(email_df.shape)

print("\nURL Dataset Shape:")
print(url_df.shape)
```

```python
print("Email Dataset Columns:")
print(email_df.columns)

print("\nURL Dataset Columns:")
print(url_df.columns)
```

```python
print(email_df['label'].value_counts())
```

```python
print(url_df['type'].value_counts())
```

```python
print("Random Email Samples")
print(email_df.sample(5))

print("\nRandom URL Samples")
print(url_df.sample(5))
```

```python
print("Missing Values in Email Dataset")
print(email_df.isnull().sum())

print("\nMissing Values in URL Dataset")
print(url_df.isnull().sum())
```

```python
sns.countplot(x='label', data=email_df)

plt.title("Phishing vs Legitimate Emails")

plt.show()
```

```python
sns.countplot(x='type', data=url_df)

plt.xticks(rotation=45)

plt.title("URL Class Distribution")

plt.show()
```
