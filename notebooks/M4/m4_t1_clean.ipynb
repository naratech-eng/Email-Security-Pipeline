# Milestone 4 - Task 1: Data Cleaning

This notebook:
- Loads the raw phishing email dataset
- Removes duplicate records using `text_combined`
- Checks class balance before and after cleaning
- Saves cleaned data to `data/processed/emails_clean.csv`

```python
from pathlib import Path

import pandas as pd
```

```python
# Build robust input/output paths from notebook location
project_root = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()

input_path = project_root / "data" / "raw" / "phishing-emails" / "phishing_email.csv"
output_dir = project_root / "data" / "processed"
output_path = output_dir / "emails_clean.csv"

print(f"Project root: {project_root}")
print(f"Input file:   {input_path}")
print(f"Output file:  {output_path}")
```

```python
# 1) Load raw dataset
df = pd.read_csv(input_path)

print(f"Columns: {df.columns.tolist()}")
print(f"Total rows loaded: {len(df)}")

print("\nClass distribution before dedup:")
class_counts_before = df["label"].value_counts()
print(class_counts_before)
```

```python
# 2) Remove duplicate emails using combined text
rows_before = len(df)
df = df.drop_duplicates(subset=["text_combined"])
rows_after = len(df)

print(f"Rows before dedup: {rows_before}")
print(f"Rows after dedup:  {rows_after}")
print(f"Duplicates removed: {rows_before - rows_after}")
```

```python
# 3) Check class balance after deduplication
print("Class distribution after dedup:")
class_counts_after = df["label"].value_counts()
print(class_counts_after)

imbalance_ratio = class_counts_after.max() / class_counts_after.min()
if imbalance_ratio > 3:
    print(f"⚠️  WARNING: imbalance ratio {imbalance_ratio:.1f}:1")
else:
    print(f"✅ Balance OK: ratio {imbalance_ratio:.1f}:1")
```

```python
# 4) Save cleaned dataset
output_dir.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)

print(f"✅ Saved cleaned dataset: {output_path} ({len(df)} rows)")
```
